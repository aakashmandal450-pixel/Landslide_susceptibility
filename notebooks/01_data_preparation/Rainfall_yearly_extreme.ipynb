```python
# --- Configuration ---
folder_path = '/content/drive/MyDrive/RemoteSensingResearchRelated/ColabNotebooks/Rainfall'  # change this to your folder path
Station = "Station_220"
input_folder = f'{folder_path}/{Station}'
print(input_folder)
```

```python

# --- Initialize output list ---
extreme_values = []
RainMax = {}

# --- Loop through all files ---
for filename in os.listdir(input_folder):
    Rain = []
    year = ''.join(filter(str.isdigit, filename))[-4:]  # extract numbers (assumed to be year)
    file_path = os.path.join(input_folder, filename)
    # Read file assuming 2 rows: days and rainfall
    df = pd.read_csv(file_path, header=None)
    # (day,*_, rain) = (str(df.values[1])[2:-2].strip(" ").split(" "))
    for i in df.index:
        try:
            rain = str(df.values[i])[2:-2].strip(" ").split(" ")[-1]
            # rain = float(str(df.values[i])[2:-2].strip(" ").split(" ")[1].strip(" "))
            # print(rain)
            if rain != '-99.9':
                Rain.append(float(rain))
        except Exception as e:
            print(f"Invalid Data skipped: {e}")
    Rain.sort()
    RainMax[year]=Rain[-1]
print(RainMax)
```

```python
df_ex = pd.DataFrame([RainMax])
df_ex = df_ex.transpose()
df_ex.to_csv(f'{folder_path}/{Station}.csv')
```
